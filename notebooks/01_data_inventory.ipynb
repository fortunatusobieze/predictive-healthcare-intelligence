```python
import pandas as pd
from src.data_loader import load_all_data
from src.utils import standardize_columns

data = load_all_data()

for name in data:
    data[name] = standardize_columns(data[name])

list(data.keys())
```

```python
# table level summary

summary = []
for name, df in data.items():
    summary.append({
        "table": name,
        "rows": df.shape[0],
        "columns": df.shape[1],
        "missing_cells": int(df.isna().sum().sum()),
        "duplicate_rows": int(df.duplicated().sum())
    })

inventory_df = pd.DataFrame(summary).sort_values("table")
inventory_df
```

```python
# date ranges

def get_date_ranges(df, table_name):
    date_cols = [c for c in df.columns if "date" in c or "datetime" in c]
    out = []
    for col in date_cols:
        temp = pd.to_datetime(df[col], errors="coerce")
        out.append({
            "table": table_name,
            "date_column": col,
            "min_date": temp.min(),
            "max_date": temp.max(),
            "null_dates": temp.isna().sum()
        })
    return out

date_summary = []
for name, df in data.items():
    date_summary.extend(get_date_ranges(df, name))

date_summary_df = pd.DataFrame(date_summary)
date_summary_df
```
